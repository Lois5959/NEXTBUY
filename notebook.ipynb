Importer les fichiers csv et afficher les 10 premières lignes de chaque fichier


```python
import pandas as pd

df_aisles = pd.read_csv("datasets-project/datasets/aisles.csv",sep=",", decimal=".")
df_departments = pd.read_csv("datasets-project/datasets/departments.csv",sep=",", decimal=".")
df_order_products = pd.read_csv("datasets-project/datasets/order_products.csv",sep=",", decimal=".")
df_orders = pd.read_csv("datasets-project/datasets/orders.csv",sep=",", decimal=".")
df_products = pd.read_csv("datasets-project/datasets/products.csv",sep=",", decimal=".", on_bad_lines='skip')


from IPython.display import display

print("Aisles:")
display(df_aisles.head(10))

print("Departments:")
display(df_departments.head(10))

print("Order Products:")
display(df_order_products.head(10))

print("Orders:")
display(df_orders.head(10))

print("Products:")
display(df_products.head(10))
```

Pour obtenir les dimensions du dataframe

```python
print(df_aisles.shape)
print(df_departments.shape)
print(df_order_products.shape)
print(df_orders.shape)
print(df_products.shape)
```

Suppression des doublons


```python
for df in [df_aisles, df_departments, df_products, df_orders, df_order_products]:
    df.drop_duplicates(inplace=True)
```

Enlever les espaces inutiles

```python
df_products.columns = df_products.columns.str.strip()
df_departments.columns = df_departments.columns.str.strip()
df_aisles.columns = df_aisles.columns.str.strip()
df_order_products.columns = df_order_products.columns.str.strip()
df_orders.columns = df_orders.columns.str.strip()
```

Garder que les colonnes utiles


```python
df_products_utile = df_products[['product_id', 'product_name', 'aisle_id', 'department_id']]
df_departments_utile = df_departments[['department_id', 'department']]
df_aisles_utile = df_aisles[['aisle_id', 'aisle']]
df_order_products_utile = df_order_products[['order_id', 'product_id', 'add_to_cart_order', 'reordered']]
df_orders_utile = df_orders[['order_id', 'user_id', 'order_dow', 'order_hour_of_day', 'days_since_prior_order']]
```

Combiner les dataframes

```python
newdf_catalog = df_products_utile.merge(df_departments_utile, on='department_id', how='inner')
newdf_catalog = newdf_catalog.merge(df_aisles_utile, on='aisle_id', how='inner')
df_sales = newdf_catalog.merge(df_order_products_utile, on='product_id', how='inner')
df_final = df_sales.merge(df_orders_utile, on='order_id', how='inner')

print(df_final)
```

```python
cols_to_remove = ['product_id', 'aisle_id', 'department_id']
df_final = df_final.drop(columns=cols_to_remove)

print(df_final.info())
```

Sauvegarde

```python
df_final.to_csv('df_clean.csv', index=False)
```
